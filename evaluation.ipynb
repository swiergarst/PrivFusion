```python
import pandas as pd
```

```python
def load_dataframes(id: str):
    indonesia_orig = pd.read_csv("datasets/covid_indonesia.csv")
    indonesia_checkpoints = [
        pd.read_csv(f"dataset_checkpoints/covid19-indonesia_modified_round{i}_{id}.csv") for i in range(2)
    ]
    # indonesia_final = pd.read_csv(f"dataset_checkpoints/covid19-indonesia_modified_round2_{id}.csv")

    global_orig = pd.read_csv("datasets/covid_global.csv")
    global_checkpoints = [
        pd.read_csv(f"dataset_checkpoints/covid19-global_modified_round{i}_{id}.csv") for i in range(2)
    ]
    # global_final = pd.read_csv(f"dataset_checkpoints/covid19-global_modified_round2_{id}.csv")

    return indonesia_orig, indonesia_checkpoints, global_orig, global_checkpoints
```

```python
id = 1791969010.8668191
indonesia_orig, indonesia_checkpoints, global_orig, global_checkpoints = load_dataframes(id)
```
